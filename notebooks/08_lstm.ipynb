```python

# Load clinical sequences using the existing patient split

sequences = []
labels = []
patient_split = []

splits = [
    ("train", train_ids),
    ("validation", val_ids),
    ("test", test_ids),
]

for split_name, patient_ids in splits:
    for pid in patient_ids:
        file_path = patient_map[pid]
        df = pd.read_csv(file_path, sep="|")

        if "SepsisLabel" not in df.columns:
            continue

        # Separate target and clinical features
        X = df.drop(columns=["SepsisLabel"]).copy()
        y = df["SepsisLabel"].copy()

        # Keep the same feature columns and order
        X = X.apply(pd.to_numeric, errors="coerce")
        X = X.replace([np.inf, -np.inf], np.nan)
        X = X.ffill().bfill().fillna(0)

        sequences.append(X.to_numpy(dtype=np.float32))
        labels.append(y.to_numpy(dtype=np.float32))
        patient_split.append(split_name)

    print(f"{split_name}: loaded {len(patient_ids)} patients")

print("Total sequences:", len(sequences))
print("Training sequences:", patient_split.count("train"))
print("Validation sequences:", patient_split.count("validation"))
print("Test sequences:", patient_split.count("test"))
```

```python
import numpy as np
import pandas as pd
import torch
from pathlib import Path
from torch.nn.utils.rnn import pad_sequence

# Map patient IDs to their file paths
patient_map = {
    Path(path).stem: path
    for path in patient_files
}

print("Patient files found:", len(patient_map))
print("Example patient IDs:", list(patient_map.keys())[:5])

def load_sequences(patient_ids):
    X, y = [], []

    for pid in patient_ids:
        df = pd.read_csv(patient_map[str(pid)], sep="|")

        if "SepsisLabel" not in df.columns:
            continue

        features = df.drop(columns=["SepsisLabel"])
        target = df["SepsisLabel"].to_numpy(dtype=np.float32)

        features = features.select_dtypes(include=[np.number])
        features = features.replace([np.inf, -np.inf], np.nan)
        features = features.ffill().bfill()
        features = features.fillna(0)

        X.append(torch.tensor(
            features.to_numpy(dtype=np.float32)
        ))
        y.append(torch.tensor(target))

    return X, y

X_train, y_train = load_sequences(train_ids)
X_val, y_val = load_sequences(val_ids)
X_test, y_test = load_sequences(test_ids)

print("Training patients:", len(X_train))
print("Validation patients:", len(X_val))
print("Testing patients:", len(X_test))
```

```python
import torch
from torch.nn.utils.rnn import pad_sequence

# Pad clinical feature sequences
X_train_pad = pad_sequence(X_train, batch_first=True)
X_val_pad = pad_sequence(X_val, batch_first=True)
X_test_pad = pad_sequence(X_test, batch_first=True)

# Pad corresponding labels
y_train_pad = pad_sequence(y_train, batch_first=True)
y_val_pad = pad_sequence(y_val, batch_first=True)
y_test_pad = pad_sequence(y_test, batch_first=True)

# Create masks to ignore padding
train_mask = torch.zeros_like(y_train_pad)
val_mask = torch.zeros_like(y_val_pad)
test_mask = torch.zeros_like(y_test_pad)

for i, y in enumerate(y_train):
    train_mask[i, :len(y)] = 1

for i, y in enumerate(y_val):
    val_mask[i, :len(y)] = 1

for i, y in enumerate(y_test):
    test_mask[i, :len(y)] = 1

print("Training tensor:", X_train_pad.shape)
print("Validation tensor:", X_val_pad.shape)
print("Testing tensor:", X_test_pad.shape)
print("Training time steps:", int(train_mask.sum().item()))
print("Positive training labels:",
      int((y_train_pad * train_mask).sum().item()))
```

```python

import torch
import torch.nn as nn

device = torch.device("cpu")

class SepsisLSTM(nn.Module):
    def __init__(self, input_size, hidden_size=64):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=1,
            batch_first=True
        )
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        output, _ = self.lstm(x)
        return self.fc(output).squeeze(-1)

model = SepsisLSTM(
    input_size=X_train_pad.shape[2]
).to(device)

positive = (y_train_pad * train_mask).sum()
negative = train_mask.sum() - positive
pos_weight = negative / positive.clamp(min=1)

criterion = nn.BCEWithLogitsLoss(
    pos_weight=pos_weight,
    reduction="none"
)

optimizer = torch.optim.Adam(
    model.parameters(), lr=0.001
)

print("Device:", device)
print("Training patients:", len(X_train))
print("Input features:", X_train_pad.shape[2])
print("Positive labels:", int(positive.item()))
print("Negative labels:", int(negative.item()))
print("Positive class weight:", round(pos_weight.item(), 2))
print("Model ready!")
```

```python
import torch
import torch.nn as nn
from torch.nn.utils.rnn import pad_sequence

device = torch.device("cpu")
model = model.to(device)

# Rebuild padded labels and a mask for real time steps
y_train_pad = pad_sequence(y_train, batch_first=True)
y_test_pad = pad_sequence(y_test, batch_first=True)

train_mask = torch.zeros_like(y_train_pad)
test_mask = torch.zeros_like(y_test_pad)

for i, y in enumerate(y_train):
    train_mask[i, :len(y)] = 1

for i, y in enumerate(y_test):
    test_mask[i, :len(y)] = 1

# Calculate positive-class weight from training labels only
all_train_labels = torch.cat(y_train)
positives = all_train_labels.sum()
negatives = len(all_train_labels) - positives

pos_weight = negatives / positives.clamp(min=1)

criterion = nn.BCEWithLogitsLoss(
    pos_weight=pos_weight,
    reduction="none"
)

optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

print("Device:", device)
print("Positive training labels:", int(positives))
print("Negative training labels:", int(negatives))
print("Positive class weight:", round(pos_weight.item(), 2))
print("Training setup ready!")
```

```python
epochs = 10

for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()

    outputs = model(X_train_pad)

    loss_values = criterion(outputs, y_train_pad)
    loss = (loss_values * train_mask).sum() / train_mask.sum()

    loss.backward()
    optimizer.step()

    print(
        f"Epoch [{epoch + 1}/{epochs}], "
        f"Loss: {loss.item():.4f}"
    )
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
import numpy as np
import torch

model.eval()

with torch.no_grad():
    logits = model(X_test_pad)
    probabilities = torch.sigmoid(logits).cpu().numpy()

y_true = y_test_pad.cpu().numpy()
mask = test_mask.cpu().numpy().astype(bool)

# Keep only actual observations, excluding padding
y_true = y_true[mask].astype(int)
y_prob = probabilities[mask]
y_pred = (y_prob >= 0.5).astype(int)

print("Test observations:", len(y_true))
print("Positive labels:", int(y_true.sum()))
print("Predicted positives:", int(y_pred.sum()))
print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))

if len(np.unique(y_true)) == 2:
    print("AUROC:", round(roc_auc_score(y_true, y_prob), 4))
else:
    print("AUROC: Cannot calculate; test labels contain only one class.")
```

```python
import numpy as np
import pandas as pd
import torch

from torch.utils.data import Dataset, DataLoader

# Load the existing patient splits
train_ids = pd.read_csv(
    results_dir / "train_patient_ids.csv"
)["patient_id"].astype(str).tolist()

val_ids = pd.read_csv(
    results_dir / "val_patient_ids.csv"
)["patient_id"].astype(str).tolist()

test_ids = pd.read_csv(
    results_dir / "test_patient_ids.csv"
)["patient_id"].astype(str).tolist()

# Use the 40 clinical features, excluding the target
FEATURE_COLUMNS = [
    col for col in df.columns
    if col != "SepsisLabel"
]

print("Number of input features:", len(FEATURE_COLUMNS))


class SepsisDataset(Dataset):
    def __init__(self, patient_ids, patient_files):
        self.patient_ids = patient_ids
        self.patient_files = patient_files

    def __len__(self):
        return len(self.patient_ids)

    def __getitem__(self, index):
        patient_id = self.patient_ids[index]
        file_path = self.patient_files[patient_id][0]

        data = pd.read_csv(file_path, sep="|")

        # Separate clinical features and hourly labels
        X = data[FEATURE_COLUMNS].copy()
        y = data["SepsisLabel"].to_numpy(dtype=np.float32)

        # Handle infinite and missing values
        X = X.replace([np.inf, -np.inf], np.nan)
        X = X.ffill().bfill().fillna(0)

        X = X.to_numpy(dtype=np.float32)

        return (
            torch.tensor(X, dtype=torch.float32),
            torch.tensor(y, dtype=torch.float32)
        )


def collate_patients(batch):
    # Keep sequences separate, without padding the whole dataset
    features, labels = zip(*batch)
    return list(features), list(labels)


train_dataset = SepsisDataset(train_ids, patient_files)
val_dataset = SepsisDataset(val_ids, patient_files)
test_dataset = SepsisDataset(test_ids, patient_files)

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    collate_fn=collate_patients,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    collate_fn=collate_patients,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    collate_fn=collate_patients,
    num_workers=0
)

print("Training patients:", len(train_dataset))
print("Validation patients:", len(val_dataset))
print("Test patients:", len(test_dataset))
```

```python
import torch
import torch.nn as nn

class SepsisLSTM(nn.Module):
    def __init__(self, input_size=40, hidden_size=64):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=1,
            batch_first=True
        )

        self.dropout = nn.Dropout(0.2)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        output, _ = self.lstm(x)
        output = self.dropout(output)
        logits = self.fc(output).squeeze(-1)
        return logits


model = SepsisLSTM()

print(model)
```

```python
import torch
import torch.nn as nn

device = torch.device("cpu")
model = model.to(device)

# Estimate positive and negative hourly labels
positive = 0
negative = 0

for paths in patient_files.values():
    file_path = paths[0]
    data = pd.read_csv(file_path, sep="|")
    labels = data["SepsisLabel"].value_counts()

    positive += int(labels.get(1, 0))
    negative += int(labels.get(0, 0))

print("Positive hourly labels:", positive)
print("Negative hourly labels:", negative)

pos_weight = torch.tensor(
    [negative / max(positive, 1)],
    dtype=torch.float32,
    device=device
)

criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

print("Positive class weight:", pos_weight.item())
print("Optimizer: Adam")
print("Learning rate: 0.001")
```

```python
import torch
import torch.nn as nn
from torch.nn.utils.rnn import pad_sequence

def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    total_loss = 0.0
    total_patients = 0

    for X_list, y_list in loader:
        X_padded = pad_sequence(
            X_list, batch_first=True, padding_value=0.0
        ).to(device)

        y_padded = pad_sequence(
            y_list, batch_first=True, padding_value=0.0
        ).to(device)

        lengths = torch.tensor([len(y) for y in y_list])
        max_len = y_padded.size(1)

        mask = (
            torch.arange(max_len).unsqueeze(0)
            < lengths.unsqueeze(1)
        ).to(device)

        optimizer.zero_grad()

        logits = model(X_padded)

        # Calculate loss only for real hourly records
        loss = criterion(logits[mask], y_padded[mask])

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        batch_size = len(X_list)
        total_loss += loss.item() * batch_size
        total_patients += batch_size

    return total_loss / total_patients
```

```python
loss = train_one_epoch(
    model=model,
    loader=train_loader,
    optimizer=optimizer,
    criterion=criterion,
    device=device
)

print("Training loss after epoch 1:", round(loss, 4))
```

```python
import numpy as np
from sklearn.metrics import roc_auc_score

def evaluate_auc(model, loader, device):
    model.eval()
    all_labels = []
    all_probs = []

    with torch.no_grad():
        for X_list, y_list in loader:
            X_padded = pad_sequence(
                X_list, batch_first=True, padding_value=0.0
            ).to(device)

            logits = model(X_padded)
            probs = torch.sigmoid(logits).cpu().numpy()

            for i, y in enumerate(y_list):
                length = len(y)
                all_labels.extend(y.numpy().tolist())
                all_probs.extend(probs[i, :length].tolist())

    all_labels = np.asarray(all_labels)
    all_probs = np.asarray(all_probs)

    if len(np.unique(all_labels)) < 2:
        return float("nan")

    return roc_auc_score(all_labels, all_probs)
```

```python
for epoch in range(1, 6):
    train_loss = train_one_epoch(
        model, train_loader, optimizer, criterion, device
    )

    val_auc = evaluate_auc(model, val_loader, device)

    print(
        f"Epoch {epoch}/5 | "
        f"Train Loss: {train_loss:.4f} | "
        f"Validation ROC-AUC: {val_auc:.4f}"
    )
```

```python
import numpy as np
from sklearn.metrics import (
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

model.eval()

all_labels = []
all_probs = []

with torch.no_grad():
    for X_list, y_list in test_loader:
        X_padded = pad_sequence(
            X_list, batch_first=True, padding_value=0.0
        ).to(device)

        logits = model(X_padded)
        probs = torch.sigmoid(logits).cpu().numpy()

        for i, y in enumerate(y_list):
            length = len(y)
            all_labels.extend(y.numpy().tolist())
            all_probs.extend(probs[i, :length].tolist())

y_true = np.asarray(all_labels, dtype=int)
y_prob = np.asarray(all_probs)

threshold = 0.5
y_pred = (y_prob >= threshold).astype(int)

print("Test ROC-AUC:", round(roc_auc_score(y_true, y_prob), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))

tn, fp, fn, tp = confusion_matrix(
    y_true, y_pred, labels=[0, 1]
).ravel()

print("\nTrue positives:", tp)
print("False positives:", fp)
print("False negatives:", fn)
print("True negatives:", tn)
print("False alarm rate:", round(fp / (fp + tn), 4))
```

```python
import numpy as np
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

# Get predictions from validation patients
model.eval()
val_labels = []
val_probs = []

with torch.no_grad():
    for X_list, y_list in val_loader:
        X_padded = pad_sequence(
            X_list, batch_first=True, padding_value=0.0
        ).to(device)

        logits = model(X_padded)
        probs = torch.sigmoid(logits).cpu().numpy()

        for i, y in enumerate(y_list):
            length = len(y)
            val_labels.extend(y.numpy().tolist())
            val_probs.extend(probs[i, :length].tolist())

val_labels = np.asarray(val_labels, dtype=int)
val_probs = np.asarray(val_probs)

# Compare thresholds using validation data only
print("Threshold | Precision | Recall | F1-score")

for threshold in [0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]:
    val_pred = (val_probs >= threshold).astype(int)

    precision = precision_score(
        val_labels, val_pred, zero_division=0
    )
    recall = recall_score(
        val_labels, val_pred, zero_division=0
    )
    f1 = f1_score(
        val_labels, val_pred, zero_division=0
    )

    print(
        f"{threshold:.2f}      | "
        f"{precision:.4f}    | "
        f"{recall:.4f} | "
        f"{f1:.4f}"
    )
```
